```python
#changing the data type of date column and deleting the timestamp
```

```python
df['requested_date'] = pd.to_datetime(df['requested_date'], format='%Y/%m/%d %I:%M:%S %p')
df['requested_date'] = df['requested_date'].dt.date
```

```python
df['closed_date'] = pd.to_datetime(df['closed_date'], format='%Y/%m/%d %I:%M:%S %p')
df['closed_date'] = df['closed_date'].dt.date
```

```python
#dropping unnecessary columns 
```

```python
columns_to_drop = ["service_request_id", "source", "address", "comm_code",
                   "location_type", "point", "updated_date"]
df.drop(columns_to_drop, axis=1, inplace=True)
```
